```python
import os

# Define sample temperature readings
temperature_readings = [
    {'timestamp': '2023-10-27 10:00:00', 'location': 'Room A', 'temperature_c': 22.5, 'humidity': 60},
    {'timestamp': '2023-10-27 10:05:00', 'location': 'Room B', 'temperature_c': 23.1, 'humidity': 58},
    {'timestamp': '2023-10-27 10:10:00', 'location': 'Room A', 'temperature_c': 22.7, 'humidity': 61},
    {'timestamp': '2023-10-27 10:15:00', 'location': 'Room C', 'temperature_c': 21.9, 'humidity': 65}
]

# Define the function to generate a report line
def generate_report_line(reading_dict):
    return f"Timestamp: {reading_dict['timestamp']}, Location: {reading_dict['location']}, Temp: {reading_dict['temperature_c']}°C, Humidity: {reading_dict['humidity']}%"


reports_dir = 'reports'
os.makedirs(reports_dir, exist_ok=True)

report_filename = os.path.join(reports_dir, 'temperature_report.txt')

try:
    with open(report_filename, 'w') as f:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            f.write(line + '\n')

    print(f"\nTemperature report written to '{report_filename}'.")

except IOError as e:
    print(f"\nError writing file '{report_filename}': {e}")
```

```python
# 4.2 Read the content of 'sensor_reports/temperature_report.txt' back into a list of strings
#     named 'read_report_lines'. Print the first 3 lines.

# Your code for Task 4.2 here:

# The report was written to 'reports/temperature_report.txt'
read_report_filename = os.path.join(reports_dir, 'temperature_report.txt')

read_report_lines = []
try:
    with open(read_report_filename, 'r') as f:
        read_report_lines = f.readlines()

    print("\nFirst 3 lines of the report:")
    for i, line in enumerate(read_report_lines):
        if i < 3:
            print(line.strip())
        else:
            break

except FileNotFoundError:
    print(f"Error: The file '{read_report_filename}' was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```
